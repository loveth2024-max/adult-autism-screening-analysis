###  **Comparing Screening Scores by Classification**

Comparing the number of participants and their average and median screening scores across the two classification groups.

```python
df.groupby('CLASS_ASD')['RESULT'].agg(['count', 'mean', 'median'])
```

### ** Visualising Screening Scores by Classification**

Using a boxplot to compare the distribution and spread of screening scores across the two classification groups.

```python

import matplotlib.pyplot as plt

df.boxplot(column='RESULT', by='CLASS_ASD', patch_artist=True,
           boxprops=dict(facecolor='skyblue'))

plt.title('Screening Scores by Classification')
plt.suptitle('')
plt.xlabel('Classification')
plt.ylabel('Screening Score')
plt.show()
```

### ** Comparing Screening Question Responses**

Comparing average responses to the ten screening questions across the two classification groups.

```python
df.groupby('CLASS_ASD')[[f'A{i}_SCORE' for i in range(1, 11)]].mean().T.plot(kind='bar')
plt.title('Screening Responses by Classification')
plt.show()
```

###  **Identifying Differences in Screening Question Responses**

Calculating the differences in average responses between the two classification groups to identify which screening questions show the largest differences.

```python
(df.groupby('CLASS_ASD')[[f'A{i}_SCORE' for i in range(1, 11)]].mean().diff().iloc[-1]).sort_values(ascending=False)
```

###  **Visualising Differences in Screening Question Responses**

Using a horizontal bar chart to highlight which screening questions show the largest differences between the YES and NO groups.

```python
(df.groupby('CLASS_ASD')[[f'A{i}_SCORE' for i in range(1,11)]].mean().diff().iloc[-1].sort_values()).plot(kind='barh', title='Difference in Screening Responses')
plt.show()
```

### ** Comparing Median Age by Classification**

Comparing the median ages of participants across the two screening classification groups.

```python
df.groupby('CLASS_ASD')['AGE'].median()
```

###  **Visualising Age Distribution by Classification**

Using a boxplot to compare age distributions across the two screening classification groups.

```python
df.boxplot(column='AGE', by='CLASS_ASD')
plt.title('Age by Screening Classification')
plt.suptitle('')
plt.grid(False)
plt.show()
```

### ** Analysing Age Groups by Classification**

Examining the distribution of participants across different age groups and screening classifications.

```python
pd.cut(df['AGE'], [0, 20, 30, 40, 50, 100]).groupby(df['CLASS_ASD']).value_counts()
```
